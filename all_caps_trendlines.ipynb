```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.pyplot import Axes
import seaborn as sns
from helpers import get_years_dfs
```

```python
sns.set_theme(context="paper", font_scale=2.5)
```

```python
def plot_all_caps_proportions(input_path: str, title: str, save_path: str):
    df = pd.concat(get_years_dfs(input_path))

    plt.figure(figsize=(12, 8))
    plt.title(title)

    sns.lineplot(x="year", y="proportion_upper_to_alpha_tokens", data=df, marker="o", linewidth=3, markersize=10)

    plt.xticks(df["year"].unique(), rotation=45)

    plt.ylabel("")

    plt.tight_layout()
    plt.savefig(save_path)
    plt.draw()
```

```python
plot_all_caps_proportions(
    input_path="./output/all_caps.xlsx",
    title="Proportion of uppercase to total alphabetical tokens for all text types",
    save_path="./visualization/all_caps/trendline_all_types.png"
)
```
